### Exercise 3 · the dormant field
List every milestone that happened *inside* a winter. Does the field look dead?

```python
for a, b, name in WINTERS:
    inside = df[(df.year >= a) & (df.year <= b) & (df.kind != 'shock')]
    print(f'\n{name} ({a}-{b}):')
    for _, r in inside.iterrows():
        print(f"  {r.year}  {r['name']}")
```
